#Join tables and Write to GOLD layer 

```python
%python
querry= """
select s.order_number,s.customer_id,concat(c.firstname,' ',c.lastname) as customer_name,c.gndr,s.product_key,p.product_name,
p.product_category,s.order_date,s.quantity,s.price,s.sales,s.sales_status from workspace.silver.tbl_sales_details as s
left join workspace.silver.crm_cust_info as c 
on  
s.customer_id=c.id
left join workspace.silver.tbl_prd_info as p 
on
s.product_key=p.Product_key"""
df=spark.sql(querry)
df.write.mode("overwrite").\
    format("delta").\
        option("overwriteSchema","true").\
            saveAsTable("workspace.gold.fact_sales")
```

```python
%sql
select customer_name, count(customer_name) from workspace.gold.fact_sales
group by customer_name 
having count(customer_name) > 1
```
