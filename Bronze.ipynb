#Reading Data from the volume

```python
df = spark.read.csv("/Volumes/workspace/bronze/source_system/sales_details.csv",   
header=True,
inferSchema=True)

df.display()
```

#Writing the Data to the Bronze

```python
df.write.format("delta")\
.mode("overwrite")\
.option("overwriteSchema", "true")\
.saveAsTable("workspace.bronze.sales_details")

df.display()
```
